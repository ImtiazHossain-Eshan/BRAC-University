```python
# 1_2

input= open('/content/drive/MyDrive/BRACU 5th Semester/CSE221/CSE221_Imtiaz Hossain/LAB_Solutions/Lab-02/input1_2.txt','r')
output= open('/content/drive/MyDrive/BRACU 5th Semester/CSE221/CSE221_Imtiaz Hossain/LAB_Solutions/Lab-02/output1_2.txt','w')

temp= input.readline().rstrip("\n").split()

N= int(temp[0])   #-------> Length of the list
S= int(temp[1])   #-------> The target sum

temp1= input.readline().rstrip("\n").split()
# print(temp1)
start= 0
end= N-1

flag= False

while start<end:
    if (int(temp1[start]) + int(temp1[end]))== S:
        output.writelines(f"{start+1} {end+1}")
        flag= True
        break

    elif (int(temp1[start]) + int(temp1[end])) < S:
        start+= 1

    elif (int(temp1[start]) + int(temp1[end])) > S:
        end-= 1

if flag==False:
    output.writelines(f"IMPOSSIBLE")

input.close()
output.close()
```

**Task 3**

```python
input= open('/content/drive/MyDrive/BRACU 5th Semester/CSE221/CSE221_Imtiaz Hossain/LAB_Solutions/Lab-02/input3.txt','r')
output= open('/content/drive/MyDrive/BRACU 5th Semester/CSE221/CSE221_Imtiaz Hossain/LAB_Solutions/Lab-02/output3.txt','w')

number_of_tasks= int(input.readline())
start= []
end= []

for i in range(number_of_tasks):   #--------Separating into 2 list
    temp= input.readline().rstrip("\n").split()

    start.append(int(temp[0]))
    end.append(int(temp[1]))

for i in range(number_of_tasks):   #--------Selection Sort in both
    min_idx= i
    for j in range(i+1, number_of_tasks):
        if end[min_idx] > end[j]:
            min_idx= j

    start[i], start[min_idx]= start[min_idx], start[i]
    end[i], end[min_idx]= end[min_idx], end[i]

# print(start)
# print(end)

final_start= []
final_end= []

for i in range(number_of_tasks):   #--------Creating Final Usable List
    if i==0:

        final_start.append(start[i])
        final_end.append(end[i])


    elif (start[i] >= final_end[(len(final_end))-1]):
        final_start.append(start[i])
        final_end.append(end[i])

    # print(final_start)
    # print(final_end)

# print(final_list)

output.write(f"{str(len(final_end))}\n")

for i in range(len(final_end)):
    output.write(f"{final_start[i]} ")
    output.write(f"{final_end[i]}\n")

input.close()
output.close()
```
