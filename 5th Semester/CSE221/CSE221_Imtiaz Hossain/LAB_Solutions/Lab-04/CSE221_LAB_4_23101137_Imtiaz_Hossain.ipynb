# Task 2 (BFS)

```python
from collections import deque
```

```python
input= open('/content/drive/MyDrive/BRACU 5th Semester/CSE221/CSE221_Imtiaz Hossain/LAB_Solutions/Lab-04/input2.txt','r')
output= open('/content/drive/MyDrive/BRACU 5th Semester/CSE221/CSE221_Imtiaz Hossain/LAB_Solutions/Lab-04/output2.txt','w')

vertices, total_edges = [int(i) for i in input.readline().split()]
edges= []

for i in range(total_edges):
    source, destination= [int(k) for k in input.readline().split()]
    edges.append((source, destination))

# print(edges)

################################################################################
def create_graph(edges):
    G = {}

    for u, v in edges:
        if u not in G:
            G[u] = {'adj': [], 'colour': 0}
        if v not in G:
            G[v] = {'adj': [], 'colour': 0}

        G[u]['adj'].append(v)
        G[v]['adj'].append(u)

    return G

G = create_graph(edges)
print(G)
# for vertex in G:
#     print(f"Vertex {vertex}: Adjacents: {G[vertex]['adj']}, Colour: {G[vertex]['colour']}")
################################################################################

def bfs(G, s):

    for u in G:
        G[u]['colour'] = 0  # White

    Q = deque()
    traversal_path = []


    G[s]['colour'] = 1  # Gray
    Q.append(s)


    while Q:
        u = Q.popleft()
        traversal_path.append(u)

        # For each adjacent vertex
        for v in G[u]['adj']:
            if G[v]['colour'] == 0:
                G[v]['colour'] = 1
                Q.append(v)

    return traversal_path


traversal_path= bfs(G, 1)
print(traversal_path)

# for vertex in G:
#     print(f"Vertex {vertex}: Colour {G[vertex]['colour']}")

################################################################################
for i in range(len(traversal_path)):
    output.write(str(traversal_path[i])+" ")
input.close()
output.close()
```
