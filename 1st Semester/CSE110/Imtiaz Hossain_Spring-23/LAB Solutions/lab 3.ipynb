```python
n = int(input("Enter the value of N: "))
fibonacci = [0, 1]
while fibonacci[-1] < n:
    next_fibonacci = fibonacci[-1] + fibonacci[-2]
    fibonacci.append(next_fibonacci)
print("The Fibonacci sequence up to", n, "is:")
for number in fibonacci:
    if number <= n:
        print(number)
```

```python
decimal = int(input("Enter a decimal number: "))

binary = ""

# Handle the special case of 0 separately
if decimal == 0:
    binary = "0"
else:
    # Convert the decimal number to binary
    while decimal > 0:
        remainder = decimal % 2
        binary = str(remainder) + binary
        decimal //= 2

# Print the binary representation of the decimal number
print("The binary representation of", decimal, "is", binary)
```
