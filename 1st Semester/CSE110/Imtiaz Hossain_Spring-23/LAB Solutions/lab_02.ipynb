```python
#task 22
can=int(input("Enter the number of canvas you want to buy\n"))
tub=int(input("Enter the number of paint tubes you want to buy\n"))
total=can*120 + tub*75
dis=0
if 0<=total<=299:
  dis=0
elif 300<=total<=499:
  dis=10
elif 500<=total<=749:
  dis=20
elif 750<=total<=999:
  dis=50
elif total>=1000:
  dis=150

ntotal=total-dis
print(f"Previous total : {total}")
print(f"New total after discount : {ntotal}")
```

```python
# TAsk 23
F=int(input("Enter the temperature in Fahrenheit "))
C=(F-32)*0.56
season=''
if C<20:
  season="Winter"
elif 20<=C<=25:
  season="Autumn"
elif 25<C<30:
  season="Spring"
elif C>=30:
  season="Summer"
print(f"{C} degree C")
print(f"{season}")
```
