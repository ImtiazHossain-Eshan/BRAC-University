```python
#Task 10
string=input("Enter values \n")
List=[]
num=''
for i in string:
  if i == ",":
    List.append(int(num))
    num=''
  elif i == " ":
    continue                # Try to shorten it
  else:
    num+=i
List.append(int(num))

modList=[]
check=[]
modList.append(List[0])
check.append(List[0])
for i in List:
  if i in check:
    continue
  else:
    modList.append(i)
    check.append(i)  
print(List)
print(modList)
```
