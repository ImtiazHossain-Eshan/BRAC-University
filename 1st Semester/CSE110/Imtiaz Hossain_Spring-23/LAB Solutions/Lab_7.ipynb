```python
#Task2
def fibonacci(n):
  l=[0,1]
  i=0
  while(l[-1]!=n):
    l+=[l[i]+l[i+1]]
    i+=1
    if l[-1]>n:
      l.pop(-1)
      break
  for i in l:
    print(i,end=" ")

fibonacci(10)
```

```python
# Recursion
def factorial(n):
  if n==0:
    return 1
  else:
    return n*fibonacci(n-1)

print(fibonacci(4))
```

```python
# Task 8
def show_palindrome(n):
  string=''
  for i in range(1,n+1):
    string+=str(i)+" "
  for i in range(n-1,0,-1):
    string+=str(i)+" "
  return string

def show_palindromic_triangle(n):
  for i in range(1,n+1):
    print(" "*(n-i)*2 + show_palindrome(i))
  
show_palindromic_triangle(8)
```

```python
# Task 10
def make_square(n):
  st,end = n
  dic={}
  for i in range(st , end+1):
    dic[i]=i**2
  return dic

print(make_square((5,9)))
```

```python
# Task 12
def max_2(n):
  new=[]
  count=0
  for i in n:
    if i in new:
      if new.count(i)<2:
        new.append(i)
      else:
        count+=1
    else:
      new.append(i)
  print(f"Removed: {count}")
  return new
print(max_2([1, 2, 3, 3, 3, 3, 4, 5, 8, 8]))
```

```python
#Task 13
def calc(operator, num1, num2):
  if operator=="+":
    return (num1+num2)
  elif operator=="-":
    return (num1-num2)
  elif operator=="*":
    return (num1*num2)
  elif operator=="/":
    return (num1/num2)
op=input("Enter operator: ")
n1=float(input("Enter number 1:\n"))
n2=float(input("Enter number 2:\n"))
calc(op,n1,n2)
    
```

```python
#Task 14
def bhugichugi(line, num):
  new=line[0]
  rem=""
  for i in range(1,len(line)):
    if i%num==0:
      rem+=line[i]
    else:
      new+=line[i]
  new+=rem
  return new
  
print(bhugichugi('Python is easy to learn. I love python.', 6))
```

```python
#Task 15
def thapao(items, location="Dhanmondi"):
  price={
        "Rice":105,
        "Potato":20,
        "Chicken":250,
        "Beef":510,
        "Oil":85}
  total=0
  for i in items:
    if i in price:
      total+=price[i]
  if location!="Dhanmondi":
    total+=70
  else:
    total+=30
  return total


items_list=input("Enter list: ").lstrip("[").rstrip("]").split(", ")
for i in range(len(items_list)):
  items_list[i]=items_list[i].strip("\"")

print(items_list)
thapao(items_list, "Mohakhali")
```
