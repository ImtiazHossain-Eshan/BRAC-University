```python
sitting_list = [10,30,20,70,11,15,22,16,58,100,12,56,70,80]

even_list = []
odd_list = []

for i in range(len(sitting_list)):
    if i % 2 == 0:
        even_list.append(sitting_list[i])
    else:
        odd_list.append(sitting_list[i])

even_list.sort()
odd_list.sort(reverse=True)

result_list = []

for i in range(len(sitting_list)):
    if i % 2 == 0:
        result_list.append(even_list[int(i/2)])
    else:
        result_list.append(odd_list[int(i/2)])

print(result_list)
```

```python
#Task-8
list_one =  [-10, 15, 2, 4, -4, 7, -8]

arr =  list_one

min_l = 0
min_r = 1
min_sum = arr[0] + arr[1]
for l in range (0, len(arr) - 1):
  for r in range (l + 1, len(arr)):
    sum = arr[l] + arr[r]                
    if abs(min_sum) > abs(sum):        
      min_sum = sum
      min_l = l
      min_r = r
 
print(f"The two elements whose sum is minimum are {arr[min_l]} and {arr[min_r]}")
```
