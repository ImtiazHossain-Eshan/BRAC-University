```python
a=input()
s=''
for i in range(len(a)-1):
  if a[i]==a[i+1]:
    continue
  else:
    if a[i]=="a" or a[i]=="A" or a[i]=="e" or a[i]=="E" or a[i]=="i" or a[i]=="I" or a[i]=="o" or a[i]=="O" or a[i]=="u" or a[i]=="U":
      s+=str(ord(a[i])) + a[i] + str(ord(a[i]))
    else:
      s+=a[i]
print(s,end="")
print(a[-1])

```
