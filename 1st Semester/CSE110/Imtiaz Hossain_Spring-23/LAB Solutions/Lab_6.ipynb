```python
# Task 8  creating Dictionary from input Dictionary
Dict=input("Enter a dictionary\n ").lstrip("{").rstrip("}").split(", ")
# print(Dict)
ind=0
for i in Dict:
    Dict[ind]=tuple(i.split(":"))
    ind+=1
# print(Dict)
new_dict={}
for items in Dict:
  # print(items)
  key,value = items
  key=key.strip("'")
  new_dict[key]=int(value)

count=0
sum=0
for i in new_dict.values():
  sum+=i
  count+=1
avg=sum/count
print(f"Average is {int(avg)}.")
```
