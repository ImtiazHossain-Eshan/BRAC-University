# Task 3

```python
categories = brown.categories()
print(categories)
```

```python
fiction_file_id = brown.fileids(categories = 'fiction')[0]
# raw_text = brown.raw(fileids=fiction_file_id)
raw_text = " ".join(brown.words(fileids = fiction_file_id))

fiction_token = word_tokenize(raw_text)

print(fiction_token)
```

```python
from nltk.corpus import stopwords
```

```python
stop_words = stopwords.words('english')
print(stop_words)
```

```python
filtered_words = []
for w in fiction_token:
    if w.lower() not in stop_words and w.isalpha():
        filtered_words.append(w)

print(filtered_words)
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_word = []
for word in filtered_words:
    lemmatized_word.append(lemmatizer.lemmatize(word, pos = 'v'))

print(lemmatized_word)
```

```python
word_count = FreqDist(lemmatized_word)
print(word_count.most_common(20))

word = []
count = []

for tup in word_count.most_common(20):
    word.append(tup[0])
    count.append(tup[1])

plt.figure(figsize=(12, 6))
plt.title(f"Top 20 Most Frequent Words in {fiction_file_id} (Fiction)")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.bar(word,count)
plt.tight_layout()
plt.show()
```

# Task 4

```python
gutenberg.fileids()
```

```python
austen = gutenberg.raw("austen-emma.txt")
austen_words = word_tokenize(austen)
cleaned_words = []
for w in austen_words:
    lower_w = w.lower()
    if w.isalpha() and lower_w not in stop_words:
        cleaned_words.append(lower_w)

print(cleaned_words)
```

```python
austen_word_count = FreqDist(cleaned_words)
top_10_items = austen_word_count.most_common(10)
print(top_10_items)
```

```python
from nltk.util import ngrams
```

```python
top_10_words = []
for item in top_10_items:
    top_10_words.append(item[0])

word_to_index = {}
for i, word in enumerate(top_10_words):
    word_to_index[word] = i

print("Word to Index Mapping:\n", word_to_index)

matrix = np.zeros((len(top_10_words), len(top_10_words)), dtype=int)

trigrams = list(ngrams(cleaned_words, n=3))

print(trigrams)
```

```python
for trigram in trigrams:
    for i in range(3):
        for j in range(i + 1, 3):
            word_a = trigram[i]
            word_b = trigram[j]

            if word_a in word_to_index and word_b in word_to_index:
                cw_index = word_to_index[word_a]
                nw_index = word_to_index[word_b]

                matrix[cw_index, nw_index] += 1
                matrix[nw_index, cw_index] += 1

print("Co-occurrence Matrix:\n")
print(matrix)
```

```python
spanish_text_2 = udhr.raw('Spanish_Espanol-Latin1')
print(spanish_text_2)
```

**Both of them are same, so I am working with the first one**

```python
spanish_text_1_token = word_tokenize(spanish_text_1.lower())
print(spanish_text_1_token)
```

```python
vowels = 'aeiouáéíóúü'
vowel_bigrams = []

for word in spanish_text_1_token:
    if not word.isalpha():
        continue

    sequence = []
    for i in word:
        if i in vowels:
            sequence.append(i)

    for i in range(len(sequence) - 1):
        vowel_bigrams.append("".join(sequence[i:i+2]))

fdist = FreqDist(vowel_bigrams)

result = fdist.most_common()
print(result)
```
