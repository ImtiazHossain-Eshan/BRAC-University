**Assignment Part 1: Doubly Linked List**

```python
#Assignment Part 1

class Patient:
    def __init__(self, id, name, age, bloodgroup, prev, next):
        self.id = id
        self.name = name
        self.age = age
        self.bloodgroup = bloodgroup
        self.prev= prev
        self.next= next
```

```python
class WRM:
    def __init__(self):

        self.head= Patient(None, None, None, None, None, None)

        self.head.prev= self.head
        self.head.next= self.head

        print(f"Start of a new patient list")

    def registerPatient(self,id, name, age, bloodgroup):

        temp_1= self.head.prev

        new_patient= Patient(id, name, age, bloodgroup, temp_1, self.head)

        temp_1.next= new_patient
        temp_1= temp_1.next

        self.head.prev = temp_1

        print()
        print(f"Patient {name} added successfully")


    def servePatient(self):

        serve= self.head.next

        serve.prev.next= serve.next
        serve.next.prev= serve.prev

        serve.prev= None
        serve.next= None

        print()
        print(f"Patient {serve.name} is being served now")


    def showAllPatient(self):

        temp_2=  self.head.next

        while temp_2!= self.head:
            print()
            print(temp_2.name)
            temp_2= temp_2.next

    def canDoctorGoHome(self):

        if self.head == self.head.next:
            print()
            print(f"Yes, The doctor can go home")
        else:
            print()
            print(f"Patients are waiting. Doctor can't go home")


    def cancelAll(self):

        temp_3= self.head

        self.head.prev= temp_3
        self.head.next= temp_3

        print("Canceled all appointments successfully")

    def ReverseTheLine(self):
        current_node= self.head
        last_node= self.head.prev

        while last_node.prev != self.head:
            current_node = self.head
            last_node = self.head.prev

            while last_node.prev != self.head:
                temp1 = current_node.next
                temp2 = current_node.prev

                current_node.next = temp2
                current_node.prev = temp1

                current_node = current_node.prev


        print(f"Reversed the line successfully")

```

```python
#Write a Tester Code in this cell

print("Welcome to Waiting Room Management System")

obj = WRM()

while True:

    print('\n==Choose an Option==\n\n1. Register Patient\n2. Serve Patient\n3. Show All Patient')
    print('4. Can Doctor Go Home?\n5. Cancel All Patient\n6. Reverse The Queue\n7. Exit')

    x = int(input('\nEnter Your Choice: '))
    print()

    if x==1:
        print('Registering Patient...')
        id, name, age, bloodgroup = input('Enter ID: '),input('Enter Name: '),input('Enter Age: '),input('Enter Bloodgroup: ')

        obj.registerPatient(id, name, age, bloodgroup)

    if x==2:
        obj.servePatient()

    if x==3:
        obj.showAllPatient()

    if x==4:
        obj.canDoctorGoHome()

    if x==5:
        obj.cancelAll()

    if x==6:
        obj.ReverseTheLine()

    if x==7:
        print('Thank you for using Waiting Room Management System\nExiting...')
        break

    elif x>7 or x<1:
        print('No Such Option')
```

**Assignment Part 2: Stack**

Linked List based Stack is implemented in the following cell.

```python
class Node:
  def __init__(self,elem=None,next=None):
    self.elem = elem
    self.next = next

class Stack:
  def __init__(self):
    self.__top = None

  def push(self,elem):
    nn = Node(elem,self.__top)
    self.__top = nn

  def pop(self):
    if self.__top == None:
      #print('Stack Underflow')
      return None
    e = self.__top
    self.__top = self.__top.next
    return e.elem

  def peek(self):
    if self.__top == None:
      #print('Stack Underflow')
      return None
    return self.__top.elem

  def isEmpty(self):
    return self.__top == None
```

Task 1: Diamond Count

```python
def diamond_count(stack,string):

    count= 0
    for i in range(len(string)):

        if (string[i] == "<"):
            stack.push(string[i])

        elif (string[i] == ">"):
            if stack.peek()== "<" and stack.peek != None:
                stack.pop()
                count+=1

    return count


print('Test 01')
stack = Stack()
string = '<..><.<..>> '
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 3
unittest.output_test(returned_value, 3)
print('-----------------------------------------')


print('Test 02')
stack = Stack()
string = '<<<..<......<<<<....>'
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 1
unittest.output_test(returned_value, 1)
print('-----------------------------------------')


print('Test 03')
stack = Stack()
string = '>>><...<<..>>...>...>>>'
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 3
unittest.output_test(returned_value, 3)
print('-----------------------------------------')
```

```python
def conditional_reverse(stack):
    new_stack= Stack()

    while stack.isEmpty() != True:
        if new_stack.peek() != stack.peek():
            new_stack.push(stack.pop())
        else:
            stack.pop()


    # print()
    # print(f"Top = {new_stack.peek()}")
    # print()

    return new_stack


print('Test 01')
st=Stack()
st.push(10)
st.push(10)
st.push(20)
st.push(20)
st.push(30)
st.push(10)
st.push(50)
print('Stack:')
print_stack(st)
print('------')
reversed_stack=conditional_reverse(st)
print('Conditional Reversed Stack:')
print_stack(reversed_stack) # This stack contains 50, 10, 30, 20, 10 in this order whereas top element should be 10
print('------')
```
