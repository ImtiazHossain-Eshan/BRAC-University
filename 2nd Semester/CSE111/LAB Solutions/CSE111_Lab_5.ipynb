```python
class Student:
    def __init__(self, name, ID, cgpa):
        self.__ID = ID
        self.__cgpa = cgpa
        self.name = name

    def setId(self, ID):
        self.__ID = ID

    def getId(self):
        return self.__ID

    def setCgpa(self, cgpa):
        self.__cgpa = cgpa

    def getCgpa(self):
        return self.__cgpa


class Department:
    def __init__(self, name):
        self.name = name
        self.students = []

    def addStudent(self, *args):
        for student in args:
            if student.getId() not in [s.getId() for s in self.students]:
                self.students.append(student)
                print(f"Welcome to {self.name} department, {student.name}")
            else:
                print("Student with the same ID already exists, Please try with another ID")


    def findStudent(self, ID):
        for student in self.students:
            if student.getId() == ID:
                print("Student info:")
                print(f"Student Name: {student.name}")
                print(f"ID: {student.getId()}")
                print(f"CGPA: {student.getCgpa()}")
                return
        print("Student with this ID doesn't exist, Please give a valid ID")

    def details(self):
        print(f"Department Name: {self.name}")
        print(f"Number of student:{len(self.students)}")
        print("Details of the students:")
        for student in self.students:
            print(f"Student name: {student.name}, ID: {student.getId()}, cgpa: {student.getCgpa()}")


#Drive Code

s1 = Student("Akib", 22301010, 3.29)
s2 = Student("Reza", 22101010, 3.45)
s3 = Student("Ruhan", 23101934, 4.00)
print("1==================================")
cse = Department("CSE")
cse.findStudent(22112233)
print("2==================================")
cse.addStudent(s1,s2,s3)
print("3==================================")
cse.details()
print("4==================================")
cse.findStudent(22301010)
print("5==================================")
s4 = Student("Nakib",22301010,3.22)
cse.addStudent(s4)
print("6==================================")
s4.setId(21201220)
cse.addStudent(s4)
print("7==================================")
cse.details()
print("8==================================")
s5 = Student("Sakib",22201010,2.29)
cse.addStudent(s5)
print("9==================================")
cse.details()
```

```python
class Spaceship:
    def __init__(self, name, capacity):
        self.name = name
        self.capacity = capacity
        self.current_cargo_weight = 0
        self.cargo = []

    def load_cargo(self, cargo):
        if self.current_cargo_weight + cargo.get_weight() <= self.capacity:
            self.cargo.append(cargo.get_name())
            self.current_cargo_weight += cargo.get_weight()
        else:
            print(f"Warning: Unable to load {cargo.get_name()} inside {self.name}. Exceeds capacity by {self.current_cargo_weight + cargo.get_weight() - self.capacity}.")

    def display_details(self):
        print(f"Spaceship Name: {self.name}")
        print(f"Capacity: {self.capacity}")
        print(f"Current Cargo Weight: {self.current_cargo_weight}")
        print(f"Cargo: {self.cargo}")

class Cargo:
    def __init__(self, name, weight):
        self.__name = name
        self.__weight = weight

    def get_name(self):
        return self.__name

    def get_weight(self):
        return self.__weight


# Creating spaceships
falcon = Spaceship( "Falcon" , 50000 )
apollo = Spaceship( "Apollo" , 100000 )
enterprise = Spaceship( "Enterprise" , 220000 )
print ( "1.===================================" )
# Creating cargo
gold = Cargo( "Gold" , 20000 )
platinum = Cargo( "Platinum" , 25000 )
dilithium = Cargo( "Dilithium" , 50000 )
trilithium = Cargo( "Trilithium" , 70000 )
neutronium = Cargo( "Neutronium" , 80000 )
print ( "2.===================================" )
# Loading cargo onto spaceships
falcon.load_cargo(gold)
falcon.load_cargo(platinum)
falcon.display_details()
print ( "3.===================================" )
apollo.load_cargo(gold) # Apollo will not reach its total capacity
apollo.display_details()
print ( "4.===================================" )
falcon.load_cargo(neutronium) # This should exceed Falcon's capacity
print ( "5.===================================" )
enterprise.load_cargo(dilithium)
enterprise.load_cargo(trilithium)
enterprise.load_cargo(neutronium) # This should not exceed Enterprise's capacity
enterprise.display_details()
```

```python
class Song:
    def __init__(self, title, artist="Unknown Artist", duration=0):
        self.__title= title
        self.__artist= artist
        self.__duration = duration


    def get_title(self):
        return self.__title

    def set_title(self, new_title):
        self.__title= new_title

    def get_artist(self):
        return self.__artist

    def set_artist(self, new_artist):
        self.__artist= new_artist

    def get_durtion(self):
        return self.__duration

    def set_duration(self, new_duration):
        self.__duration= new_duration


    def info(self):
        if (self.__title!="" and self.__duration>0):
            min= (self.__duration)//60
            sec= (self.__duration) - (min*60)
            return f"{self.__title} by {self.__artist} ({min}:{sec})"

        else:
            return f"Cannot create song without title and duration!"


class Playlist:
    def __init__(self, name, *song_list):
        self.__name= name
        self.__song_list= []

        for i in song_list:
            self.__song_list.append(i)

        self.__now_id= -1

    def get_name(self):
        return self.__name

    def set_title(self, new_name):
        self.__name= new_name

    def get_song_list(self):
        return self.__song_list

    def set_song_list(self, new_song_list):
        self.__song_list= new_song_list

    def get_now_id(self):
        return self.__now_id

    def set_now_id(self, new_now_id):
        self.__now_id= new_now_id



    def play_next(self):
        if len(self.__song_list) == 0:
            print("The playlist is empty.")
            return

        self.__now_id= (self.__now_id + 1) % len(self.__song_list)

        print(f"# Now playing: {self.__song_list[self.__now_id].info()}")



    def play_prev(self):
        if len(self.__song_list) == 0:
            print("The playlist is empty.")
            return

        if self.__now_id == 0:
            print("Cannot go back from the first song.")
            print(f"Now playing: {self.__song_list[self.__now_id].info()}")
            return

        self.__now_id -= 1
        print(f"Now playing: {self.__song_list[self.__now_id].info()}")

    def add_songs(self, *songs):
        for song in songs:
            self.__song_list.append(song)

    def remove_song(self, song_obj):
        self.__song_list.remove(song_obj)

    def show_playlist(self):
        count=0

        print(f"Playlist: {self.__name}")
        print(f"Total Songs: {len(self.__song_list)}")
        for i in self.__song_list:
            count+=1
            print(f"{count}. {i.info()})")


#Driver Code

s1 = Song("The Flower Duet", "Leo Delibes", 86)
print(s1.info())
# Output:
# The Flower Duet by Leo Delibes (1:26)

print('===============')
s2 = Song("Winter")
print(s2.info())
# Output:
# Cannot create song without title and duration!
# --- Error --- by --- Error --- (0:0)

print('===============')
s2.set_title("Winter")
s2.set_duration(223)
s2.set_artist("Vivaldi")
print(s2.info())
# Output:
# Winter by Vivaldi (3:43)

print('===============')
s3 = Song("Ride of the Valkyries", "Wagner", 137)
s4 = Song("Russian Dance - Nutcracker", duration=71)

p1 = Playlist("Classical Orchestra", s1, s2, s3)
p1.show_playlist()
# Output:
# Playlist: Classical Orchestra
# Total Songs: 3
# 1. The Flower Duet by Leo Delibes (1:26)
# 2. Winter by Vivaldi (3:43)
# 3. Ride of the Valkyries by Wagner (2:17)

print('===============')
p1.play_next()
# Output:
# Now playing: The Flower Duet by Leo Delibes (1:26)

print('===============')
p1.play_prev()
# Output:
# Cannot go to previous from the first song.
# Now playing: The Flower Duet by Leo Delibes (1:26)

print('===============')
p1.play_next()
# Output:
# Now playing: Winter by Vivaldi (3:43)

print('===============')
p1.play_next()
# Output:
# Now playing: Ride of the Valkyries by Wagner (2:17)

print('===============')
p1.play_next()
# Output:
# Reached the end of the playlist, starting from the first song.
# Now playing: The Flower Duet by Leo Delibes (1:26)

print('===============')
p1.play_next()
# Output:
# Now playing: Winter by Vivaldi (3:43)

print('===============')
p1.add_songs(s4)
# Output:
# Added 1 song(s) to the playlist Classical Orchestra

print('===============')
p1.play_next()
# Output:
# Now playing: Ride of the Valkyries by Wagner (2:17)

print('===============')
p1.play_next()
# Output:
# Now playing: Russian Dance - Nutcracker by Unknown Artist (1:11)

print('===============')
p1.play_prev()
# Output:
# Now playing: Ride of the Valkyries by Wagner (2:17)

print('===============')
p1.show_playlist()
# Output:
# Playlist: Classical Orchestra
# Total Songs: 4
# 1. The Flower Duet by Leo Delibes (1:26)
# 2. Winter by Vivaldi (3:43)
# 3. Ride of the Valkyries by Wagner (2:17)
# 4. Russian Dance - Nutcracker by Unknown Artist (1:11)
```

```python
class Song:
    def __init__(self, title, duration, artist="Unknown Artist"):
        self.__title = title
        self.__artist = artist
        self.__duration = duration

    def info(self):
        min, sec = divmod(self.__duration, 60)
        return f"{self.__title} by {self.__artist} ({min}:{sec:02d})"

    def get_title(self):
        return self.__title

    def set_title(self, title):
        self.__title = title

    def get_artist(self):
        return self.__artist

    def set_artist(self, artist):
        self.__artist = artist

    def get_duration(self):
        return self.__duration

    def set_duration(self, duration):
        self.__duration = duration


class Playlist:
    def __init__(self, name, song_list=[]):
        self.__name = name
        self.__song_list = song_list
        self.__now_id = -1

    def play_next(self):
        if len(self.__song_list) == 0:
            print("The playlist is empty.")
            return
        self.__now_id = (self.__now_id + 1) % len(self.__song_list)
        print(f"Now playing: {self.__song_list[self.__now_id].info()}")

    def play_prev(self):
        if len(self.__song_list) == 0:
            print("The playlist is empty.")
            return
        if self.__now_id == 0:
            print("Cannot go back from the first song.")
            print(f"Now playing: {self.__song_list[self.__now_id].info()}")
            return
        self.__now_id -= 1
        print(f"Now playing: {self.__song_list[self.__now_id].info()}")

    def add_songs(self, *songs):
        for song in songs:
            self.__song_list.append(song)

    def remove_song(self, song_obj):
        if song_obj in self.__song_list:
            self.__song_list.remove(song_obj)

    def show_playlist(self):
        print(f"Playlist: {self.__name}")
        print(f"Total Songs: {len(self.__song_list)}")
        for i, song in enumerate(self.__song_list):
            print(f"{i+1}. {song.info()}")

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_song_list(self):
        return self.__song_list

    def set_song_list(self, song_list):
        self.__song_list = song_list

    def get_now_id(self):
        return self.__now_id

    def set_now_id(self, now_id):
        self.__now_id = now_id



#Drive Code
# Create songs
song1 = Song("a", 1, "WIZ")
song2 = Song("c", 2, "LIPA")
song3 = Song("e", 3, "BOYS")

# Create a playlist
playlist1 = Playlist("g")

# Add songs to the playlist
playlist1.add_songs(song1, song2, song3)

# Show the playlist
playlist1.show_playlist()

# Play the next song
playlist1.play_next()

# Play the next song
playlist1.play_next()

# Play the previous song
playlist1.play_prev()

# Remove a song from the playlist
playlist1.remove_song(song2)

# Show the updated playlist
playlist1.show_playlist()
```

```python
class Book:
    def __init__(self, title, author, genre):
        self.__title= title
        self.__author= author
        self.__genre= genre
        self.__available= True
        self.__borrower= None

        self.__borrower_id= None

    def set_title(self, title):
        self.__title= title
    def get_title(self):
        return self.__title

    def set_author(self, author):
        self.__author= author
    def get_author(self):
        return self.__author

    def set_genre(self, genre):
        self.__genre= genre
    def get_genre(self):
        return self.__genre

    def set_availability(self, available= True):
        self.__available= available

    def get_availability(self):
        return self.__available

    def set_borrower(self, borrower= None, borrower_id= None):
        self.__borrower= borrower
        self.__borrower_id= borrower_id
    def get_borrower(self):
        return self.__borrower

    def display_info(self):

        print(f"Title: {self.get_title()}")
        print(f"Author: {self.get_author()}")
        print(f"Genre: {self.get_genre()}")
        print(f"Availability status: {self.get_availability()}")
        print(f"Borrowed by: {self.__borrower_id}")
        print(f"---------------")


class LibraryMember:
    def __init__(self, member_id, name, borrowed_books= None):
        self.__member_id= member_id
        self.__name= name
        self.__borrowed_books= borrowed_books if borrowed_books else []
        self.__borrowed_books_user_Id_list= []

    def set_member_id(self, new_member_id):
        self.__member_id= new_member_id
    def get_member_id(self):
        return self.__member_id

    def set_name(self, new_name):
        self.__name= new_name
    def get_name(self):
        return self.__name

    def borrow_book(self, book):
        self.__borrowed_books.append(book)
        self.__borrowed_books_user_Id_list.append(self.__member_id)
        book.set_availability(False)
        book.set_borrower(self.__name, self.__member_id)

    def return_book(self, book):
        if book in self.__borrowed_books:
            self.__borrowed_books.remove(book)
            book.set_availability(True)
            book.set_borrower(None)

        else:
            print(f"The book '{book.get_title()}' is not borrowed by {self.get_name()}")

    def display_borrowed_books(self):
        print(f"Books borrowed by {self.get_name()}")
        for i in self.__borrowed_books:
            print(f"Title: {i.get_title()}")
            print(f"Author: {i.get_author()}")
            print(f"Genre: {i.get_genre()}")
            print(f"Available: {i.get_availability()}")
            print(f"---------------")

class Library:
    def __init__(self, books_available = None , library_members= None):
        self.__books_available= books_available if books_available else []
        self.__library_members= library_members if library_members else []

    def add_book(self, book):
        self.__books_available.append(book)

    def add_library_member(self, member):
        self.__library_members.append(member)

    def display_book_list(self):
        print("All the books in the library are:")
        for book in self.__books_available:
            book.display_info()

    def display_library_members(self):
        print("Library Members:")
        for member in self.__library_members:
            print(member.get_name())

#Driver Code

book1= Book("The Hobbit", "J. R. R. Tolkien", "Novel")
book2= Book("Dream of the Red Chamber", "Cao Xueqin", "Fiction")
book3 = Book("Calculus", "Gilbert Strang", "Education")

m1= LibraryMember("IH23101137", "Imtiaz")
m2= LibraryMember("ES23101138", "Eshan")

library= Library()

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

library.add_library_member(m1)
library.add_library_member(m2)

m1.borrow_book(book1)
m1.borrow_book(book2)
m2.borrow_book(book3)

library.display_book_list()

print("1======================================")

m1.display_borrowed_books()

print("2======================================")

m2.display_borrowed_books()

print("3======================================")

m1.return_book(book2)

print("4======================================")

m1.display_borrowed_books()

library.display_book_list()

```
