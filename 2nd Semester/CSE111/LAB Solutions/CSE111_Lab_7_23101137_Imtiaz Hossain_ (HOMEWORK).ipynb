```python
import datetime

class Employee:
    employee_count={}
    total=0

    def __init__(self,name, joining_date,  work_experience, weekly_work_hour=40):
        self.name= name
        self.work_experience= work_experience

        if weekly_work_hour<61:
            self.weekly_work_hour = weekly_work_hour
        else:
            print(f"The weekly work hour should not exceed 60 hours.")
            self.weekly_work_hour = 40

        self.joining_date= joining_date
        self.year, self.month, self.date = self.joining_date.split("-")

    @classmethod
    def showDetails(cls):
        print("Company workforce")
        print(f"Total employee : {Employee.total}")
        for key,value in Employee.employee_count.items():
            print(f"Total {key} Employees : {value}")

class Programmer(Employee):

    designation_list=[ "Junior Software Engineer", 'Software Engineer', 'Senior Software Engineer', 'Technical Lead']

    def __init__(self, name, joining_date, work_experience, weekly_work_hour= 40):
        super().__init__(name, joining_date, work_experience, weekly_work_hour)

        self.salary=0
        Employee.total+=1

        if "programmer" not in Employee.employee_count:
          Employee.employee_count["programmer"]=1
        else:
          Employee.employee_count["programmer"]+=1

        self.id=self.createProgrammerID()

        if 0 <= self.work_experience < 3:
           self.designation= Programmer.designation_list[0]
        elif 3 <= self.work_experience < 5:
            self.designation= Programmer.designation_list[1]
        elif 5 <= self.work_experience < 8:
            self.designation= Programmer.designation_list[2]
        elif self.work_experience>8:
             self.designation= Programmer.designation_list[3]


    def showProgrammerDetails(self):
        print("Programmer employee:")
        print(f"Name: {self.name}")
        print(f"Id: {self.id}")
        print(f"Joining date: {self.date}-{self.month}-{self.year}")
        print(f"Designation: {self.designation}")
        print(f"Salary: BDT {round(self.salary, 1)}")

    def calculateSalary(self):

        date=datetime.datetime.now()
        c_year=date.year
        year=c_year-int(self.year)
        s=0

        if self.weekly_work_hour<40:
            print(f"{self.name} can not work for {self.weekly_work_hour} hours.")

            if self.designation == Programmer.designation_list[0]:
                s+=30000
                self.salary=s*1.15**year
            elif self.designation == Programmer.designation_list[1]:
                s+= 45000
                self.salary=s*(1.15)**year
            elif self.designation == Programmer.designation_list[2]:
                s+= 70000
                self.salary=s*(1.15)**year
            else:
                s+= 120000
                self.salary=s*(1.15)**year

        else:
            if self.designation == 'Junior Software Engineer':
                s+= 30000
                self.salary=s*(1.15)**year
            elif self.designation == 'Software Engineer':
                s+= 45000
                self.salary=s*(1.15)**year
            elif self.designation == 'Senior Software Engineer':
                s+= 70000
                self.salary=s*(1.15)**year
            else:
                s+= 120000
                self.salary=s*(1.15)**year


    def createProgrammerID(self):

        st=""
        date=self.joining_date[2::]

        for i in date:
           if i!="-":
            st+=i
        f=f"P-{st}-{Employee.total}"
        return f

    def calculateOvertime(self):

        extra_hour= self.weekly_work_hour-40

        if extra_hour>0:
            overtime=extra_hour*4*500
            self.salary+=overtime
            print(f"{self.name} will get BDT {overtime} overtime.")
        else:
            print(f"{self.name} will not get overtime.")

class HR(Employee):

    def __init__(self,name,joining_date,work_experience,weekly_work_hour=40):
       super().__init__(name,joining_date,work_experience,weekly_work_hour)
       Employee.total+=1

       if "HR" not in Employee.employee_count:
          Employee.employee_count["HR"]=1
       else:
          Employee.employee_count["HR"]+=1

       self.id = self.createHREmployeeID()

    def showHREmployeeDetails(self):

      print("HR Employee:")
      print(f"Name: {self.name}")
      print(f"Id: {self.id}")
      print(f"Joining date: {self.date}-{self.month}-{self.year}")

    def createHREmployeeID(self):

        st=""
        date=self.joining_date[2::]
        for i in date:
           if i!="-":
            st+=i
        f=f"HR-{st}-{Employee.total}"
        return f


class InternProgrammer(Programmer):
     intern_count = 0

     def __init__(self,name, joining_date, intern_type='Unpaid'):
        super().__init__(name, joining_date, 0, 40)
        Employee.total-=1

        if "programmer" not in Employee.employee_count:
          Employee.employee_count["programmer"]=1
        else:
          Employee.employee_count["programmer"]-=1

        InternProgrammer.intern_count += 1

        self.intern_type=intern_type
        self.id=f"Temp_{InternProgrammer.intern_count}"

        date1="2023-08-21"
        date2=self.joining_date

        y1,m1,d1=date1.split("-")
        y2,m2,d2=date2.split("-")

        ygap=int(y1)-int(y2)-1
        ygap=ygap*12
        months=ygap+(12-int(m2))+int(m1)

        if int(months)>4:
            self.status=True
        else:
            self.status=False

     def showInternDetails(self):
        print(f"Intern Programmer\nName:", self.name)
        print("Temporary ID:", self.id)
        print(f"Joining Date:, {self.joining_date}")
        print("Type:", self.intern_type)

        if self.status:
            print("Status: Eligible for promotion")
        else:
            print("Status: Not Eligible for promotion")

     def promoteToProgrammer(self):
        self.joining_date='2023-08-12'
        if self.status:
            print(f"{self.name} is promoted!")
            return Programmer(self.name,self.joining_date, 0, 40)
        else:
            print(f"{self.name} cannot be promoted.")


Employee.showDetails()
print("=========1=========")
richard = Programmer("Richard Hendricks", "2021-06-08", 4, 48)
richard.calculateSalary()
print("=========2=========")
richard.showProgrammerDetails()
print("=========3=========")
richard.calculateOvertime()
print("=========4=========")
richard.showProgrammerDetails()
print("=========5=========")
monica = HR("Monica Hall", "2022-07-06", 2, 40)
print("=========6=========")
monica.showHREmployeeDetails()
print("=========7=========")
Employee.showDetails()
print("=========8=========")
gilfoyle = Programmer("Bertram Gilfoyle", "2020-03-02", 6, 35)
gilfoyle.calculateSalary()
print("=========9=========")
gilfoyle.calculateOvertime()
print("=========10=========")
gilfoyle.showProgrammerDetails()
print("=========11=========")
gavin = Programmer("Gavin Belson", "2016-12-20", 9)
gavin.calculateSalary()
gavin.calculateOvertime()
gavin.showProgrammerDetails()
print("=========12=========")
yang = InternProgrammer("Jian Yang", "2023-01-01")
yang.showInternDetails()
print("=========13=========")
jared = InternProgrammer("Jared Dunn", "2023-06-05", "Paid")
jared.showInternDetails()
print("=========14=========")
jared = jared.promoteToProgrammer()
print("=========15=========")
Employee.showDetails()
print("=========16=========")
yang = yang.promoteToProgrammer()
yang.calculateSalary()
yang.showProgrammerDetails()
print("=========17=========")
Employee.showDetails()
```
