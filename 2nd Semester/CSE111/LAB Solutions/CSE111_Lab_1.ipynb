**Task 1**

```python
number_list=[]
code_is_running = True

while code_is_running:
    number = input()
    if number == "STOP":
        code_is_running = False
    else:
        number_list.append(int(number))

unique_list = []

for num in number_list:
    if num not in unique_list:
        unique_list.append(num)

for num in unique_list:
    count=0
    for common_numbers in number_list:
        if num == common_numbers:
            count+=1
    
    print(f"{num} - {count} times")
```

```python
while True:
    user_inp = input().split()
    if user_inp[0]== "STOP":
        break
    
    num_list=[]

    for num in user_inp:           #[1, 4, 2, 3]
        num_list.append(int(num))

    is_ub_jumper = True
    for idx in range(len(num_list)-1):
        diff = abs(num_list[idx+1] - num_list[idx])

        if diff not in range(1, len(num_list)):
            is_ub_jumper = False
            break
    
    if is_ub_jumper:
        print("UB Jumper")
    
    else:
        print("Not UB Jumper")
    
```

**Task 1**

```python
dict_1 = input().split(",")
dict_2 = input().split(",")

new_dict = {}

for items in dict_1:
    item_strip = items.strip()
    key, value = item_strip.split(": ")
    new_dict[key] = int(value)

for items in dict_2:
    item_strip = items.strip()
    key, value = item_strip.split(": ")

    if key in new_dict:
      new_dict[key] += int(value)
    else:
      new_dict[key] = int(value)

value_list = []

for value in new_dict.values():
    if value not in value_list:
        value_list.append(value)

n = len(value_list)

for i in range(n-1):
    min_idx = i
    for j in range(i+1, n):
        if value_list[j] < value_list[min_idx]:
            min_idx = j

    value_list[i], value_list[min_idx] = value_list[min_idx], value_list[i]

print(new_dict)
print(f"Values: {tuple(value_list)}")
```
