```python
class SentenceAnalyzer:
    def __init__(self, str1=""):
        self.str1= str1

    def set_sentence(self, str1):
        self.str1= str1

    def get_word_count(self, character_count=None):

        list1=self.str1.split(" ")

        length= len(list1)

        if character_count is None:
            print(f"Number of words in the sentence: {length}")
        else:
            count= 0
            for word in list1:
                if len(word) == character_count:
                    count+=1
            print(f"Number of words with {character_count} characters in the sentence: {count}")


#Driver Code [No change]
analyzer1 = SentenceAnalyzer()


analyzer1.set_sentence("That's an easy one")
print("1--------------------------")
analyzer1.get_word_count()
print("2--------------------------")
analyzer2 = SentenceAnalyzer("Like I said it's easy")


print("3--------------------------")
analyzer2.get_word_count()


print("4--------------------------")
analyzer2.get_word_count(4)
print("5--------------------------")
analyzer1.get_word_count(5)
```

```python
class Author:
    def __init__(self, author_name=""):
        self.author_name= author_name
        self.books= {}

    def setName(self, author_name):
        self.author_name= author_name

    def addBook(self, book_name, genre):
        if self.author_name == "":
            print(f"A book can not be added without author name")

        else:
            if genre not in self.books:
                self.books[genre]= []

            self.books[genre].append(book_name)


    def printDetail(self):

        Number_of_Books=0

        for v in self.books.values():
            Number_of_Books+= len(v)

        print(f"Number of Book(s): {Number_of_Books}")
        print(f"Author Name: {self.author_name}")

        for genre, book_name in self.books.items():
            print(genre + ":", ", ".join(book_name))


#Driver Code [No change]

a1 = Author()
print("=================================")
a1.addBook("Ice", "Science Fiction")
print("=================================")
a1.setName("Anna Kavan")
a1.addBook("Ice", "Science Fiction")
a1.printDetail()
print("=================================")
a2 = Author("Humayun Ahmed")
a2.addBook("Onnobhubon","Science Fiction")
a2.addBook("Megher Upor Bari", "Horror")
print("=================================")
a2.printDetail()
a2.addBook("Ireena", "Science Fiction")
print("=================================")
a2.printDetail()
print("=================================")
```

```python
class Farmer:
    def __init__(self, inp=None):
        self.inp= inp

        self.list_corp= []
        self.list_fish= []

        if inp == None:
            print(f"Welcome to your farm!")
        elif type(inp)== int:
            print(f"Welcome to your farm. Your farm ID is {self.inp}!")
        elif type(inp)== str:
            print(f"Welcome to your farm!, {self.inp}!")

    def addCrops(self, *corp_tuple):

        for i in corp_tuple:
            self.list_corp.append(i)

        if len(corp_tuple)!=0:
            print(f"{len(corp_tuple)} crop(s) added.")
        else:
            print(f"No crop(s) added.")

    def addFishes(self, *fish_tuple):

        for i in fish_tuple:
            self.list_fish.append(i)

        if len(fish_tuple)==0:
            print(f"No fish added.")
        else:
            print(f"{len(fish_tuple)} fish(s) added.")


    def showGoods(self):
        if len(self.list_corp)>0:
            print(f"You have {len(self.list_corp)} crop(s):")
            list_corp1= ','.join(self.list_corp)
            print(f"{list_corp1}")
        else:
            print(f"You don't have any crop(s).")

        if len(self.list_fish)>0:
            print(f"You have {len(self.list_fish)} fish(s):")
            list_fish1= ','.join(self.list_fish)
            print(f"{list_fish1}")
        else:
            print(f"You don't have any fish(s).")


#Driver Code

f1 = Farmer()
print("-------------------")
f1.addCrops('Rice', "Jute", "Cinnamon")
print("-------------------")
f1.addFishes()
print("-------------------")
f1.addCrops('Mustard')
print("-------------------")
f1.showGoods()
print("-------------------")
f2 = Farmer("Korim Mia")
print("-------------------")
f2.addFishes('Pangash', 'Magur')
print("-------------------")
f2.addCrops("Wheat", "Potato")
print("-------------------")
f2.addFishes("Koi", "Tuna", "Sardine")
print("-------------------")
f2.showGoods()
print("-------------------")
f3 = Farmer(2865127000)
print("-------------------")
f3.addCrops()
print("-------------------")
f3.addFishes("Katla")
print("-------------------")
f3.showGoods()
print("-------------------")
```

```python
class Department:
    def __init__(self, dept="ChE Department", sect=5):
        self.dept= dept
        self.sect= sect
        self.students_number_list = []

        self.average= 0
        self.count= 0

        self.total_students_of_mega_department= 0
        self.average_students_of_mega_department= 0

        print(f"The {dept} has {sect} sections.")

    def add_students(self, *nums_tuple):
        for i in nums_tuple:
            self.students_number_list.append(i)

        if len(self.students_number_list) == self.sect:
            self.average = round((sum(self.students_number_list) / self.sect),2)
            print(f"The {self.dept} has an average of {self.average} students in each section.")

        else:
            print(f"The {self.dept} doesn't have {len(nums_tuple)} sections.")

    def merge_Department(self, *dept_location_tuple):
        for i in dept_location_tuple:
            self.total_students_of_mega_department+= i.average * i.sect
            print(f"{i.dept} Department is merged to {self.dept}.")

        if self.count==0:
            self.total_students_of_mega_department+= self.average * self.sect
            self.count+=1

        self.average_students_of_mega_department= round((self.total_students_of_mega_department/
                                                self.sect),2)

        return (f"Now the {self.dept} has an average of {self.average_students_of_mega_department} students in each section.")

#Driver Code
d1 = Department()
print('1-----------------------------------')
d2 = Department('MME Department')
print('2-----------------------------------')
d3 = Department('NCE Department',8)
print('3-----------------------------------')
d1.add_students(12,23,12,34,21)
print('4-----------------------------------')
d2.add_students(40,30,21)
print('5-----------------------------------')
d3.add_students(12,34,41,17,30,22,32,51)
print('6-----------------------------------')
mega = Department('Engineering Department',10)
print('7-----------------------------------')
mega.add_students(21,30,40,36,10,32,27,51,45,15)
print('8-----------------------------------')
print(mega.merge_Department(d1,d2))
print('9-----------------------------------')
print(mega.merge_Department(d3))
```

```python
class StudentRoutineGenerator:
    def __init__(self, name="", max_course=0):
        self.name= name
        self.max_course= max_course
        self.num_courses = 0

        self.all_info_list= []
        self.routine = {'Sat/Thurs': {}, 'Sun/Tue': {}, 'Mon/Wed': {}}

        print(f"Name: {self.name}")
        print(f"Maximum Number of Courses: {self.max_course}")
        print(f"Initial Routine: {self.routine}")


    def addCourses(self, *courses):
        for i in courses:

            course_code, day, time= i.split("-")

            count_1= False
            count_2= False

            if self.num_courses < self.max_course:
                for j in self.all_info_list:
                    if course_code in j[0]:
                        count_1= True
                    if day in j[1] and time in j[2]:
                        clash_Course= j[0]
                        count_2=True

                if count_1== True:
                    print(f"You already have {course_code} in your routine")

                elif count_2== True:
                    print(f"Can't take {course_code}. It's clashing with your {clash_Course}")

                else:
                    self.all_info_list.extend([(course_code, day, time)])
                    self.num_courses+= 1
                    print(f"Successfully added {course_code}!")

            else:
                print(f"You can't take more than {self.max_course} courses")


    def dropCourse(self, *drop):
        not_in= True

        for i in drop:
            for j in self.all_info_list:
                if j[0]==i:
                    not_in= False
                    self.all_info_list.remove(j)
                    print(f"Successfully dropped {i}")

            if not_in== True:
                print(f"No such course in your routine")

    def showRoutine(self):

        self.routine = {'Sat/Thurs': {}, 'Sun/Tue': {}, 'Mon/Wed': {}}

        for k, v in self.routine.items():
            for i in self.all_info_list:
                if i[1]==k:
                    v[i[2]]=i[0]

        print(f"Updated Routine:")
        print(self.routine)

        print(f"Routine Details:")

        for k, v in self.routine.items():
            if len(v)!=0:
                print(f"{k}:")
                for m,n in v.items():
                    print(f"{m}-{n}")

#Driver Code

print('########################################')
st1 = StudentRoutineGenerator('Harry', 4)
print('------------------------------')
st1.addCourses('CSE110-Mon/Wed-12:30', 'MAT110-Mon/Wed-2:00')
st1.addCourses('ENG101-Sun/Tue-12:30', 'CSE110-Mon/Wed-9:30')
st1.addCourses('PHY111-Sun/Tue-12:30')
print('------------------------------')
st1.showRoutine()
print('------------------------------')
st1.dropCourse('CSE110')
st1.dropCourse('PHY112')
print('------------------------------')
st1.showRoutine()
print('########################################')
st2 = StudentRoutineGenerator('John', 3)
print('------------------------------')
st2.addCourses('MAT110-Mon/Wed-8:00')
st2.addCourses('ENG101-Sat/Thurs-12:30', 'CSE110-Sun/Tue-9:30')
st2.addCourses('PHY111-Sun/Tue-12:30')
print('------------------------------')
st2.showRoutine()
```
