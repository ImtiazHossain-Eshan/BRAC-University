```python
def food_order(burger_name, place= "Mohakhali"):

    meal_cost=0
    delivery_charge= 0

    if burger_name == "BBQ Chicken Cheese Burger":
        meal_cost+= 250

    elif burger_name == "Beef Burger":
        meal_cost+= 170

    elif burger_name == "Naga Drums":
        meal_cost+= 200

    tax= (meal_cost)*(8/100)

    if place == "Mohakhali":
        delivery_charge+= 40

    else:
        delivery_charge+= 60

    Total_Price = meal_cost + delivery_charge + tax

    return Total_Price

print(food_order("Beef Burger", "Dhanmondi"))
```

```python
def palindrome_checker(str1):

    str2=""
    str3=""

    length1= len(str1)   #10

    for i in range(0,length1):
        if str1[i]!= " ":
            str2+= str1[i]

    str3= str2[::-1]

    if str2==str3:
        result= "Palindrome"
    else:
        result= "Not a Palindrome"

    return result

print(palindrome_checker("nurses run"))
```

```python
def days_calculator(num):

    years= num//365
    months= (num-(years*365)) // 30
    days= num-(months*30 + years*365)

    result= f"{years} years, {months} months, {days} days"
    return result

print(days_calculator(4000))
```

```python
def grammarly(line):

    line=line.capitalize()
    chk=['.','?','!']
    line=[i for i in line]

    for i in range(len(line)):
        if line[i] in chk and i<len(line)-1:
            line[i+2]=line[i+2].upper()
        if (line[i]=='i' and line[i-1]==" " and line[i+1]==" "
                                            and i<len(line)-1):
            line[i]="I"

    return "".join(line)

print(grammarly('my favourite animal is a dog. a dog has sharp teeth so that it can eat flesh very easily. do you know my pet dog’s name? i love my pet very much.'))
```
