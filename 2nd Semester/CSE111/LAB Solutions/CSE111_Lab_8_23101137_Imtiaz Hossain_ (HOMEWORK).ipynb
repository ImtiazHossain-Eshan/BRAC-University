```python
class Exam:
    def __init__(self,marks):
        self.marks = marks
        self.time = 60

    def examSyllabus(self):
        return "Maths , English"
    def examParts(self):
        return "Part 1 - Maths\nPart 2 - English\n"



class ScienceExam(Exam):
    def __init__(self, marks, time, *subs):
        super().__init__(marks)

        self.time= time
        self.sub_dict= {}
        self.exam_parts= 2
        self.s= ""
        self.sy= ""

        for i in subs:
            self.exam_parts= self.exam_parts+1
            self.sub_dict[self.exam_parts]= i

    def examParts(self):
        for k, v in self.sub_dict.items():
            self.s+= f"Part {k} - {v}\n"

        return super().examParts() + self.s

    def examSyllabus(self):
        for v in self.sub_dict.values():
            self.sy+= f" , {v}"

        return super().examSyllabus() + self.sy

    def __str__(self):
        return f"Marks: {self.marks} Time: {self.time} minutes Number of Parts: {self.exam_parts}"


engineering = ScienceExam(100,90,"Physics","HigherMaths")
print(engineering)
print('----------------------------------')
print(engineering.examSyllabus())
print(engineering.examParts())
print('==================================')
architecture = ScienceExam(100,120,"Physics","HigherMaths","Drawing")
print(architecture)
print('----------------------------------')
print(architecture.examSyllabus())
print(architecture.examParts())
```

```python
class PokemonBasic:

  def __init__(self, name = 'Default', hp = 0, weakness = 'None', type = 'Unknown'):
    self.name = name
    self.hit_point = hp
    self.weakness = weakness
    self.type = type

  def get_type(self):
    return 'Main type: ' + self.type

  def get_move(self):
    return 'Basic move: ' + 'Quick Attack'

  def __str__(self):
    return "Name: " + self.name + ", HP: " + str(self.hit_point) + ", Weakness: " + self.weakness


class PokemonExtra(PokemonBasic):
    def __init__(self, name = 'Default', hp = 0, weakness = 'None', type = 'Unknown', secondary_type="", other_move=None):
        super().__init__(name, hp, weakness, type)

        self.secondary_type= secondary_type
        self.s= other_move

    def get_type(self):
        if self.secondary_type != "":
            return super().get_type() + f", Secondary type: {self.secondary_type}"
        else:
            return super().get_type()

    def get_move(self):
        if self.s:
            return super().get_move() + f"\nOther move: {', '.join(self.s)}"
        else:
            return super().get_move()



print('\n------------Basic Info:--------------')
pk = PokemonBasic()
print(pk)
print(pk.get_type())
print(pk.get_move())

print('\n------------Pokemon 1 Info:-------------')
charmander = PokemonExtra('Charmander', 39, 'Water', 'Fire')
print(charmander)
print(charmander.get_type())
print(charmander.get_move())

print('\n------------Pokemon 2 Info:-------------')
charizard = PokemonExtra('Charizard', 78, 'Water', 'Fire', 'Flying', ('Fire Spin', 'Fire Blaze'))
print(charizard)
print(charizard.get_type())
print(charizard.get_move())
```

```python
class Cakes:
    order_list= {}
    feedback_dict={}

    def __init__(self, typ, weight):
        self.typ= typ
        self.weight= weight

        self.price= (self.weight/1000)*1200

        self.sweetness= "Moderate"
        self.cream_type= "Whipped Cream"

        self.count= 0


        if self.typ not in Cakes.order_list:
            s= f"{self.typ} Cake {self.weight}gm"
            Cakes.order_list[s]= self.count+1

        else:
            s= f"{self.typ} Cake {self.weight}gm"
            self.order_list[s].append(self.count+1)


    def cake_details(self):
        print(f"Flavor: {self.typ} Cake, Weight: {self.weight} gm")
        print(f"Sweetness: {self.sweetness} sugar, Cream Type: {self.cream_type}")
        print(f"Price: {self.price} Taka")


    def add_customization(self, sweetness1, cream1):
        self.sweetness= sweetness1
        self.cream_type= cream1

    @classmethod
    def give_feedbacks(cls, typ1, cmnt):

        if typ1 not in cls.feedback_dict:
            cls.feedback_dict[typ1]= [cmnt]

        else:
            cls.feedback_dict[typ1].append(cmnt)


        print(f"Thanks for your valuable feedback!")

    @classmethod
    def show_feedbacks(cls):
        print(cls.feedback_dict)



class Cheese_Cakes(Cakes):

    discount_token = 0
    limit=1

    def __init__(self, typ, weight, cake_type="baked"):
        super().__init__(typ, weight)


        if self.discount_token == 0:
            self.price = (self.weight / 1000) * 1500

        elif self.discount_token == 1:
            self.price = (self.weight / 1000) * 1500 - 0.10 * ((self.weight / 1000) * 1500)
            Cheese_Cakes.discount_token -= 1

        self.cream_type= "Cream Cheese"
        self.cake_type= cake_type


    def cake_details(self):
        super().cake_details()
        c= f"Cake Type: {self.cake_type}"

        print(c)

    def add_customization(self):
        print(f"Sorry! No customization available for cheese cakes.")

    @classmethod
    def give_feedbacks(cls, typ1, cmnt):
        super().give_feedbacks(typ1, cmnt)

        print(f"You will get 10% discount for your next purchase!")

        if cls.discount_token < cls.limit:
            cls.discount_token+=1
        else:
            pass


#Driver Code

order_1=Cakes("Chocolate",500)
order_2=Cakes("Vanilla",800)
print("(1)**********************************")
order_1.cake_details()
print("(1.1)**********************************")
print(Cakes.order_list)
print("(2)**********************************")
order_2.add_customization("Zero","Butter Cream")
order_2.cake_details()
print("(3)**********************************")
Cakes.give_feedbacks("Chocolate Cake","Very Delicious")
Cakes.give_feedbacks("Chocolate Cake","Yummy")
print("(4)**********************************")
Cakes.show_feedbacks()
print("(5)**********************************")
ch_order1=Cheese_Cakes("Red velvet",700)
ch_order1.cake_details()
print("(6)**********************************")
ch_order1.add_customization()
print("(7)**********************************")
ch_order2=Cheese_Cakes("Blue Berry",900,"No Bake")
ch_order2.cake_details()
print("(8)**********************************")
print(Cakes.order_list)
print("(9)**********************************")
Cheese_Cakes.give_feedbacks("Red velvet Cheese Cake","Average")
Cheese_Cakes.give_feedbacks("Blue Berry Cheese Cake","Liked it")
print("(10)**********************************")
Cakes.show_feedbacks()



# Extra- [Code to check Discount] (For Practice)

# print("(11)**********************************")
# ch_order4=Cheese_Cakes("Red velvet",700)
# ch_order4.cake_details()
# print("(12)**********************************")
# ch_order5=Cheese_Cakes("Red velvet",700)
# ch_order5.cake_details()
# print("(13)**********************************")
# ch_order6=Cheese_Cakes("Red velvet",700)
# ch_order6.cake_details()
# print("(14)**********************************")
# print(Cheese_Cakes.discount_token)
# print("(15)**********************************")
# Cheese_Cakes.give_feedbacks("Red velvet Cheese Cake","Average")
# print("(9)**********************************")
# ch_order7=Cheese_Cakes("Red velvet",700)
# ch_order7.cake_details()
```
